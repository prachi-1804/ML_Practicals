```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv('retail_sales_dataset.csv')
```

```python
required_cols=['Transaction ID','Date','Customer ID','Gender','Age','Product Category','Quantity','Prize Per Unit','Total Amount']
```

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
df.head()
```

```python
df.info()
```

```python
duplicates = df.duplicated().sum()
if duplicates > 0:
  df.drop_duplicates(inplace=True)
  print(f"dropped {duplicates} duplicates rows")
```

```python
df['Transaction Id']=df['Transaction ID'].astype('int64')
df['Customer Id']=df['Customer ID'].astype('string')
df['Gender']=df['Gender'].astype('category')
df['Product Category']=df['Product Category'].astype('category')
print("\n data cleaning complete")
```

Categorial Graph

```python
df.groupby('Product Category')['Total Amount'].sum().plot(kind='bar')
plt.title("Sales by Category")
plt.show()
```

```python

df['Product Category'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.show()
```
