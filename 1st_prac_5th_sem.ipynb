Feature Engineering

```python
# Extract Month Name

df['Month_Name'] = df['Date'].dt.month_name()

df.head()
```

```python
# Extract Day Name

df['Day_Name'] = df['Date'].dt.day_name()

df.head()
```

```python
# Create Age Group

df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0,20,35,50,100],
    labels=['Teen','Young Adult','Adult','Senior']
)

df.head()
```

```python
# Create Purchase Category

df['Purchase_Category'] = pd.cut(
    df['Total Amount'],
    bins=[0,300,700,1500],
    labels=['Low','Medium','High']
)

df.head()
```

```python
# Create Customer Segment

df['Customer_Segment'] = df['Total Amount'].apply(
    lambda x: 'Premium' if x >= 700 else 'Regular'
)

df.head()
```

```python
# Create Weekend Feature

df['Weekend'] = df['Day_Name'].apply(
    lambda x: 'Weekend' if x in ['Saturday','Sunday'] else 'Weekday'
)

df.head()
```

```python
# Save Updated Dataset

df.to_csv('updated_retail_sales_dataset.csv', index=False)

print("Updated dataset saved successfully.")
```
